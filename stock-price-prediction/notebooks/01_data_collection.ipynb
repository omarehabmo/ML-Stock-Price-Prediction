```python
#Create Moving Averages

df['MA7'] = df['Close'].rolling(window=7).mean()

df['MA30'] = df['Close'].rolling(window=30).mean()

df['MA90'] = df['Close'].rolling(window=90).mean()

df[['Close', 'MA7', 'MA30', 'MA90']].tail()
```

```python
df['Lag_1'] = df['Close'].shift(1)

df['Lag_2'] = df['Close'].shift(2)

df['Lag_3'] = df['Close'].shift(3)
df[['Close', 'Lag_1', 'Lag_2', 'Lag_3']].head(10)
```

```python
df['Price_Range'] = df['High'] - df['Low']
df[['High', 'Low', 'Price_Range']].head()
```

```python
df['Open_Close_Diff'] = df['Close'] - df['Open']
```

```python
df['Daily_Return'] = df['Close'].pct_change()
```

```python
df.isnull().sum()
df = df.dropna()
df.isnull().sum()
```

```python
df.head()
df.info()
```

```python
df.to_csv("../data/processed/apple_stock_features.csv")
```

## Feature Engineering

The following features were created to improve the predictive capability of the machine learning models:

- 7-day moving average (MA7)
- 30-day moving average (MA30)
- 90-day moving average (MA90)
- Daily return
- 30-day rolling volatility
- Lag features (1, 2, and 3 days)
- Daily price range
- Open-close price difference

Missing values introduced by rolling windows and lag operations were removed before modeling.

```python
df.columns
```

```python
X = df.drop(columns=['Close'])

y = df['Close']
print(X.shape)
print(y.shape)
```

```python
split_index = int(len(df) * 0.8)

X_train = X[:split_index]
X_test = X[split_index:]

y_train = y[:split_index]
y_test = y[split_index:]
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(X_train, y_train)
predictions = model.predict(X_test)
predictions[:10]
```

```python
results = y_test.to_frame(name='Actual')

results['Predicted'] = predictions

results.head(10)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(14,6))

plt.plot(results.index, results['Actual'], label='Actual')
plt.plot(results.index, results['Predicted'], label='Predicted')

plt.title('Actual vs Predicted Closing Price')
plt.xlabel('Date')
plt.ylabel('Price ($)')
plt.legend()
plt.grid(True)

plt.show()
```

```python
results.head(10)
```

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import numpy as np
```

```python
mae = mean_absolute_error(y_test, predictions)

mse = mean_squared_error(y_test, predictions)

rmse = np.sqrt(mse)

r2 = r2_score(y_test, predictions)
```

```python
print(f"MAE  : {mae:.2f}")
print(f"MSE  : {mse:.2f}")
print(f"RMSE : {rmse:.2f}")
print(f"R²   : {r2:.4f}")
```

## Model Evaluation

The Linear Regression model was evaluated using MAE, RMSE, and R² score.

These metrics measure:

- Prediction accuracy
- Average prediction error
- Overall model performance

The results will serve as a baseline for comparison with more advanced machine learning models.
