```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))
plt.plot(df.index, df['Close'])
plt.title('Apple Stock Closing Price (2020-2024)')
plt.xlabel('Date')
plt.ylabel('Price ($)')
plt.grid(True)
plt.show()
```

```python
df.columns = df.columns.get_level_values(0)

df.columns
df.isnull().sum()
df.duplicated().sum()
```

```python
df['MA30'] = df['Close'].rolling(window=30).mean()
plt.figure(figsize=(12,6))

plt.plot(df.index, df['Close'], label='Close Price')
plt.plot(df.index, df['MA30'], label='30-Day Moving Average')

plt.title('Apple Stock Price vs 30-Day Moving Average')
plt.xlabel('Date')
plt.ylabel('Price ($)')
plt.legend()
plt.grid(True)

plt.show()
```

### Observation

The Apple stock price shows an overall upward trend from 2020 to 2024. The 30-day moving average smooths short-term fluctuations and highlights the long-term trend.
